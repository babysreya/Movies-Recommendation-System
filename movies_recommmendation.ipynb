```python
import ast
```

```python
def genre_extract(col):
    L=[]
    for i in ast.literal_eval(col):
        L.append(i['name'])
    return L
```

```python
data['genres']=data['genres'].apply(genre_extract)
```

```python
data.head(3)
```

```python
def keyword_extract(col):
    L=[]
    for i in ast.literal_eval(col):
        L.append(i['name'])
    return L
```

```python
data['keywords']=data['keywords'].apply(keyword_extract)
```

```python
data['keywords'].head(3)
```

```python
data['release_date']=pd.to_datetime(data['release_date']) 
```

```python
data['release_date']=data['release_date'].dt.year
```

```python
data['release_date'].head(3)
```

```python
def cast_extract(col):
    L=[]
    j=1
    for i in ast.literal_eval(col):
        if j<=3:
            L.append(i['name'])
            j+=1
    return L
```

```python
data['cast']=data['cast'].apply(cast_extract)
```

```python
data['cast'].head(3)
```

```python
def dir_extract(col):
    L=[]
    for i in ast.literal_eval(col):
        if i['job']=='Director':
            L.append(i['name'])
            break
    return L
    
```

```python
data['crew']=data['crew'].apply(dir_extract)
```

```python
data.head(3)
```

```python
data['overview']=data['overview'].apply(lambda x:x.split())
```

```python
data.head(3)
```

```python
data['cast']=data['cast'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
data['crew']=data['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
data['genres']=data['genres'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
data['keywords']=data['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
data.head(4)
```

```python
data['tags']=data['overview']+data['genres']+data['keywords']+data['cast']+data['crew']
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps=PorterStemmer()
```

```python
data['tags'][0]
```

```python
def stemming(text):
    L=[]
    for i in text.split():
        L.append(ps.stem(i))
    return " ".join(L)
```

```python
data['tags']=data['tags'].apply(stemming)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors=cv.fit_transform(data['tags'])
```

```python
# vectors=vectors.toarray()
```

```python
vectors
```

```python
cv.get_feature_names_out()[:200]
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarities=cosine_similarity(vectors)
```

```python
similarities
```

```python
similarities[0]
```

```python
data[data['title']=='avatar'].index[0]
```

```python
def recommend(movie):
    movie_index=data[data['title']==movie].index[0]
    recommendations=similarities[movie_index]
    movie_list=sorted(enumerate(recommendations),reverse=True,key=lambda x:x[1])[1:6]
    for i in movie_list:
        print(data.iloc[i[0]].title)
    return recommendations
```
